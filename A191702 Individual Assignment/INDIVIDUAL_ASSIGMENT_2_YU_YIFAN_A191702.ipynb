```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Assignment 1 dataset
df = pd.read_csv("shopify_sales_products.csv")

# Select required attributes
df = df[['Price_Clean', 'Inventory_Clean']]

# Convert to numeric
df['Price_Clean'] = pd.to_numeric(df['Price_Clean'], errors='coerce')
df['Inventory_Clean'] = pd.to_numeric(df['Inventory_Clean'], errors='coerce')

# Drop missing values
df.dropna(inplace=True)

df.head()
```

## Histogram & Boxplot (Price_Clean)

Histogram

```python
plt.figure()
plt.hist(df['Price_Clean'], bins=15)
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.title("Histogram of Product Prices")
plt.show()
```

Boxplot

```python
plt.figure()
plt.boxplot(df['Price_Clean'])
plt.ylabel("Price")
plt.title("Boxplot of Product Prices")
plt.show()
```

Minimum & Maximum

```python
min_price = df['Price_Clean'].min()
max_price = df['Price_Clean'].max()

min_price, max_price
```

## Report Explanation

The histogram shows that most products are priced within the lower to middle price range, indicating that the majority of items are moderately priced consumer products.

The boxplot reveals a right-skewed distribution, where a small number of products have significantly higher prices compared to the rest of the dataset.

The minimum price represents entry-level products, while the maximum price corresponds to premium products.

Outliers are observed in the boxplot as points above the upper whisker. These outliers represent high-priced products that differ considerably from the majority.

## Scatterplot & Correlation
## (Price_Clean vs Inventory_Clean)

Scatterplot

```python
plt.figure()
plt.scatter(df['Price_Clean'], df['Inventory_Clean'])
plt.xlabel("Price")
plt.ylabel("Inventory Quantity")
plt.title("Scatterplot of Price vs Inventory")
plt.show()
```

Correlation

```python
correlation = df['Price_Clean'].corr(df['Inventory_Clean'])
correlation
```

## Report Explanation
The scatterplot illustrates the relationship between product price and inventory quantity.

The distribution of points suggests a weak negative relationship, which is confirmed by the correlation coefficient value. This indicates that higher-priced products tend to have lower inventory levels, while lower-priced products are often stocked in larger quantities.

Several isolated data points can be identified as outliers, such as products with high prices but unusually high inventory levels. These outliers may represent special promotions or bulk-stocked premium items.
